# Robot learning course

Start with the new course and follow the notebook order. Each unit has a learner notebook and a reference notebook in the same folder. The notebook contains its exercise. Save your own copy to keep your answers.

The course assumes Python and NumPy. PyTorch is explained when first needed. CPU checks run by default. Longer GPU experiments are explicit and use local assets; notebooks do not download assets or start long jobs automatically.

## Core course

- **Playground:** [01-playground](01-playground/01-playground.ipynb) · [reference](01-playground/01-playground-reference.ipynb) — Run working ACT, diffusion, and VLA policies; inspect observations, actions, and resets.
- **Execution:** [02-run-policy](02-run-policy/02-run-policy.ipynb) · [reference](02-run-policy/02-run-policy-reference.ipynb) — Implement the episode loop and connect learner actions to the simulator.
- **ACT training:** [03-act-training](03-act-training/03-act-training.ipynb) · [reference](03-act-training/03-act-training-reference.ipynb) — Implement ACT losses, an optimizer update, and a bounded training loop.
- **Adapters:** [04-adapters](04-adapters/04-adapters.ipynb) · [reference](04-adapters/04-adapters-reference.ipynb) — Convert camera, state, and action values at the policy/controller boundary.
- **Data:** [05-data](05-data/05-data.ipynb) · [reference](05-data/05-data-reference.ipynb) — Read episodes, build windows, split whole episodes, and compute training-only statistics.
- **CVAE:** [06-cvae](06-cvae/06-cvae.ipynb) · [reference](06-cvae/06-cvae-reference.ipynb) — Implement CVAE sampling and latent loss operations.
- **Attention:** [07-attention](07-attention/07-attention.ipynb) · [reference](07-attention/07-attention-reference.ipynb) — Implement attention operations with explicit shapes and masks.
- **ACT decoder:** [08a-act-decoder](08a-decoder/08a-decoder.ipynb) · [reference](08a-decoder/08a-decoder-reference.ipynb) — Implement decoder operations.
- **ACT assembly:** [08b-act-assembly](08b-act/08b-act.ipynb) · [reference](08b-act/08b-act-reference.ipynb) — Connect ACT components and check fixed-weight parity.
- **Diffusion training:** [09-diffusion-training](09-diffusion-training/09-diffusion-training.ipynb) · [reference](09-diffusion-training/09-diffusion-training-reference.ipynb) — Implement the objective and connect it to a real update.
- **Denoiser blocks:** [10a-denoiser-blocks](10a-denoiser-blocks/10a-denoiser-blocks.ipynb) · [reference](10a-denoiser-blocks/10a-denoiser-blocks-reference.ipynb) — Implement denoiser building blocks.
- **U-Net:** [10b-unet](10b-unet/10b-unet.ipynb) · [reference](10b-unet/10b-unet-reference.ipynb) — Assemble and check the U-Net path.
- **Sampling:** [11-sampling](11-diffusion-sampling/11-diffusion-sampling.ipynb) · [reference](11-diffusion-sampling/11-diffusion-sampling-reference.ipynb) — Implement sampling steps and the executed action slice.
- **VLA inputs:** [12a-vla-inputs](12a-vla-inputs/12a-vla-inputs.ipynb) · [reference](12a-vla-inputs/12a-vla-inputs-reference.ipynb) — Prepare images, state, tokens, and masks for a VLA.
- **Flow matching:** [12b-flow](12b-flow/12b-flow.ipynb) · [reference](12b-flow/12b-flow-reference.ipynb) — Implement flow pairs and loss; verify frozen weights.
- **Data improvement:** [13-data-quality](13-data-quality/13-data-quality.ipynb) · [reference](13-data-quality/13-data-quality-reference.ipynb) — Improve alignment and data quality; preserve episode boundaries.
- **Comparison:** [14-comparison](14-comparison/14-comparison.ipynb) · [reference](14-comparison/14-comparison-reference.ipynb) — Pair results on fixed task, state, and seed cases; compare methods.

Short function exercises are about 15 minutes. Larger experiments take longer. These are learning targets, not measured times. Use states 0–9 for exploration, 10–14 for method checks, 15–19 for data-intervention checks, and 20–49 for final evaluation after Unit 13 and recipe freeze. Keep training/validation episode splits separate from these state groups. Record task, partition, seed, checkpoint provenance, and timing with each experiment.

# Robot learning course

This course starts with a live ACT policy in LIBERO-Goal. Labs 0 to 11 are available on the prepared local machine. The revised keyboard controls passed a user retest. The paid cloud path remains open. Run commands from the repository root in the locked environment.

1. [Lab 0: Run ACT](00-first-run/00-first-run.ipynb) — load a supplied checkpoint, run a task, watch a video, and reset. About 1–2 learner hours.
2. [Lab 1: Understand ACT](01-act/01-act.ipynb) — inspect camera/state tensors, the CVAE loss, action chunks, and temporal ensembling. About 2–3 learner hours.
3. [Lab 2: Data and evaluation](02-data-evaluation/02-data-evaluation.ipynb) — audit a complete demonstration, the train-only processor, and fixed live evaluation states. About 3–4 learner hours.
4. [Lab 3: Train ACT](03-train-act/03-train-act.ipynb) — submit a training probe, resume a checkpoint, and compare demonstration counts and action chunks. About 3–5 learner hours.
5. [Lab 4: Run SmolVLA](04-smolvla/04-smolvla.ipynb) — inspect the published policy and test instructions on a fixed goal. About 2–3 learner hours.
6. [Lab 5: Adapt SmolVLA](05-adapt-smolvla/05-adapt-smolvla.ipynb) — trace VLM initialization, train C1 and C2, and compare seeded development results. About 4–5 learner hours.
7. [Lab 6: Diffusion Policy](06-diffusion/06-diffusion.ipynb) — check data windows, train and reload a policy, then compare denoising steps and live results. About 3–4 learner hours.
8. [Lab 7: Compare methods](07-comparison/07-comparison.ipynb) — inspect the frozen paired final results on two tasks and choose a practical recipe. About 3–4 learner hours.
9. [Lab 8: Data quality and recovery](08-data-quality/08-data-quality.ipynb) — check action timing, repair a labelled fault, record actions, and assess a bounded ACT continuation. About 3–4 learner hours.
10. [Lab 9: Runtime faults](09-runtime/09-runtime.ipynb) — step delayed chunks on virtual ticks and check outages, underflow, and reset. About 2–3 learner hours.
11. [Lab 10: RTC mechanism](10-rtc/10-rtc.ipynb) — check the mask and guidance derivative, then read the frozen five-mode result. About 4–6 learner hours.
12. [Lab 11: Operations](11-deployment/11-deployment.ipynb) — check checkpoint provenance, test a rollback pointer, and write a career memo. About 2–3 learner hours.

Run `.venv/bin/robot-lab doctor --live` before Lab 0 and `.venv/bin/robot-lab data fetch --scope training` before Labs 1 to 6. Run `.venv/bin/robot-lab data fetch-models` before Labs 4 to 6. Use final simulator states 20 to 49 only with the frozen Lab 7 and Lab 10 recipes. The [course plan](../docs/plans/vla-tutorial-plan.md) lists the gates. The [operations guide](../docs/operations.md) gives restore and rollback steps.

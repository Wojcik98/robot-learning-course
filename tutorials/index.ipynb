# Robot learning course

This course starts with a live ACT policy in LIBERO-Goal. Labs 0 and 1 are ready on the prepared local machine. Later data, SmolVLA, Diffusion Policy, and timing labs are in progress. Run commands from the repository root in the locked environment.

1. [Lab 0: Run ACT](00-first-run/00-first-run.ipynb) — load a supplied checkpoint, run a task, watch a video, and reset. About 1–2 learner hours.
2. [Lab 1: Understand ACT](01-act/01-act.ipynb) — inspect camera/state tensors, the CVAE loss, action chunks, and temporal ensembling. About 2–3 learner hours.

Run `.venv/bin/robot-lab doctor --live` before Lab 0 and `.venv/bin/robot-lab data fetch --scope training` before Lab 1. Keep final simulator states 20 to 49 closed until final evaluation. The [course plan](../docs/plans/vla-tutorial-plan.md) lists the later labs and gates.

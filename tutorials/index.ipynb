# Robot learning course

This course starts with working policies and teaches you to build and train three models for robot control: ACT (action chunking with transformers), diffusion, and SmolVLA (a small vision-language-action policy). You implement small functions for execution, training, data, and model mechanics in 22 lessons, each with a learner and reference notebook.

Start with the playgrounds. Work through ACT, then diffusion, then SmolVLA. Each lesson has its own exercise. Required simulator and training cells are separate from small CPU examples. Each training experiment targets a 48 GB GPU and a maximum of two hours. The quick automation profile skips these experiment cells; the full profile executes them.

Before you start, follow the repository [setup guide](../docs/initial-setup.md) and [environment guide](../docs/environment.md). Check live model and simulator assets with `just doctor-live`.

## How to work through a lesson

Each lesson folder has two notebooks:
- **Learner notebook**: exercises with places to write your answers and test cells.
- **Reference notebook**: complete answers for review.

Open the learner notebook first. **Save your own copy before you work** so your answers stay available. You do not need to use one shared kernel; later lessons import tested functions from earlier ones.

Read the contract above each function: it states inputs, shapes, units, and outputs. Implement your function, run the check cell, then use it in that lesson's experiment. Hints are in collapsed sections. Move to the reference notebook after you try the exercise.

**Data splits:** States 0–9 are for exploration. States 10–14 are for method checks. States 15–19 are for data-intervention checks (starting in Lesson 21). States 20–49 are for final evaluation after Lesson 21 and recipe freeze. Do not use final results to tune methods. Keep all attempts in every denominator. Record task, partition, seed, checkpoint provenance, and timing with each experiment.

**Shared task:** All three policies use LIBERO `bowl_plate`. Control runs at 20 Hz: each command advances one 50 ms tick. The source dataset reports a 10 FPS storage label. This label selects stored frames; it does not cause action repetition or removal of alternate rows. New recordings preserve every control transition. Saved processors define each model's feature normalization; image scaling and controller clipping are separate operations.

**Kernels:** Fresh kernels help. Each notebook's workspace discovery code lets you start from inside the lesson folder ("lab"), the tutorials folder ("tutorials"), or the repository root (".").

Start with a working ACT policy. Then build the full pipeline: execution, data, training, and the model parts (CVAE, attention, decoder, assembly, adaptation).

8. **Assembly** — [learner](08-act-assembly/08-act-assembly.ipynb) · [reference](08-act-assembly/08-act-assembly-reference.ipynb)

9. **Adaptation** — [learner](09-act-adaptation/09-act-adaptation.ipynb) · [reference](09-act-adaptation/09-act-adaptation-reference.ipynb)

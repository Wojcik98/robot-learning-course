# Robot learning course

This course starts with a live ACT policy in LIBERO-Goal. Labs 0 to 3 run on the prepared local machine. Later SmolVLA, Diffusion Policy, and timing labs are in progress. Run commands from the repository root in the locked environment.

1. [Lab 0: Run ACT](00-first-run/00-first-run.ipynb) — load a supplied checkpoint, run a task, watch a video, and reset. About 1–2 learner hours.
2. [Lab 1: Understand ACT](01-act/01-act.ipynb) — inspect camera/state tensors, the CVAE loss, action chunks, and temporal ensembling. About 2–3 learner hours.
3. [Lab 2: Data and evaluation](02-data-evaluation/02-data-evaluation.ipynb) — audit a complete demonstration, the train-only processor, and fixed live evaluation states. About 3–4 learner hours.
4. [Lab 3: Train ACT](03-train-act/03-train-act.ipynb) — submit a training probe, resume a checkpoint, and compare demonstration counts and action chunks. About 3–5 learner hours.

Run `.venv/bin/robot-lab doctor --live` before Lab 0 and `.venv/bin/robot-lab data fetch --scope training` before Labs 1 to 3. Keep final simulator states 20 to 49 closed until final evaluation. The [course plan](../docs/plans/vla-tutorial-plan.md) lists the later labs and gates.

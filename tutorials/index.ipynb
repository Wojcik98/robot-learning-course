# Robot learning course

This course starts with a live ACT policy in LIBERO-Goal. The later ACT, data, SmolVLA, Diffusion Policy, and timing labs are in progress. Run all commands from the repository root in the locked environment.

1. [Lab 0: Run ACT](00-first-run/00-first-run.ipynb) — load a supplied checkpoint, run a task, watch a video, and reset.
2. [Lab 0 instructions](00-first-run/README.md) — setup, terminal command, and result limits.
3. [Lab 0 exercise](00-first-run/exercise.md) — compare two development start states.

Run `.venv/bin/robot-lab doctor --live` before Lab 0. Keep final simulator states 20 to 49 closed until final evaluation. The [course plan](../docs/plans/vla-tutorial-plan.md) lists all planned labs and gates.

## The idea before the implementation

**Tell the denoiser what it sees and how noisy the actions are.** The same noisy action values can mean different things near different objects. The denoiser needs observation context. It also needs the noise level: a nearly clean sample calls for a different correction from a highly corrupted sample.

**Where you work.** You work inside one U-Net pass. A sinusoidal embedding represents the noise level. Observation features and that embedding form a condition vector. FiLM uses the condition to scale and shift convolution features. A residual path carries input features around the learned branch. These operations predict noise; the scheduler later uses that prediction to update the action window.

## Background: how a convolutional denoiser gets its conditions

The denoiser is a 1D convolutional network over the *time axis of the action window* (16 steps). It receives two kinds of side information:

1. **How noisy is the input?** The integer timestep `t` is turned into a smooth vector (`sinusoidal_timestep_embedding`) and passed through a small MLP.
2. **What does the robot see?** The observation features (state, camera features) form one *global condition* vector.

The two are concatenated into a `cond` vector. Every residual block turns `cond` into per-channel numbers that *modulate* the convolution features. This is **FiLM** (feature-wise linear modulation):

$$F'_{c,\tau}=\gamma_c(\text{cond})\,F_{c,\tau}+\beta_c(\text{cond})$$

Where:
* $F_{c,\tau}$ - convolution feature of channel $c$ at action-window slot $\tau$.
* $\gamma_c(\text{cond})$ - the scale for channel $c$, computed from `cond` by a learned layer.
* $\beta_c(\text{cond})$ - the bias for channel $c$, computed the same way. (This $\beta$ is not the noise schedule of Unit 11.)
* $F'_{c,\tau}$ - the modulated feature.

The same scale and bias apply to every slot $\tau$ of a channel. In the pinned policy `use_film_scale_modulation=True`, so both scale and bias are used.
### Worked example: FiLM on two channels

Suppose one time slot has convolution features $F=[1,\,1]$, FiLM scale $\gamma=[2,\,0.5]$, and bias $\beta=[-1,\,1]$. Element-wise $\gamma F+\beta$ gives $[1,\,1.5]$. The same channel values broadcast across all action-window slots. For embedding dimension 4 at timestep 0, the sine half is $[0,\,0]$ and cosine half is $[1,\,1]$; a learned MLP then turns this code into the scale and bias.


## Exercise 12.1 — Sinusoidal timestep embedding

**Purpose.** Give the network a smooth, bounded code for the noise level, instead of a raw integer.

**Contract (shapes, units, masks).**

- `sinusoidal_timestep_embedding(timesteps, dim)`: `timesteps (B,)` integer or float tensor; `dim` even and at least 4. Return `(B, dim)` `float32`.
- Let $n$ = `dim // 2`. Each row is built from $n$ frequencies, **all sines first, then all cosines** (not interleaved):

$$f_k=\exp\!\left(-\frac{k\,\ln 10000}{n-1}\right), \qquad e(t)=\left[\sin(tf_0),\dots,\sin(tf_{n-1}),\ \cos(tf_0),\dots,\cos(tf_{n-1})\right]$$

  Where:
  * $t$ - one timestep value from `timesteps`.
  * $k$ - frequency index, $0 \le k < n$; $f_0=1$ is the fastest and $f_{n-1}=1/10000$ the slowest.
  * $e(t)$ - the output row of length `dim`.
- Timestep 0 gives `sin = 0` and `cos = 1` everywhere.

**Method and source.** The Transformer position code applied to the diffusion timestep ([DDPM](https://arxiv.org/abs/2006.11239), appendix B; [Vaswani et al. 2017](https://arxiv.org/abs/1706.03762), 3.5). The layout differs from Exercise 6.4 (that one interleaves sin and cos and counts frequencies differently).

**Pointers.** Make the frequency vector first. Multiply the column of timesteps with the frequency row to get `(B, half)` angles. Concatenate the sines and the cosines on the last axis.

<details><summary>Detailed hint</summary>

```python
half = dim // 2
scale = math.log(10000) / (half - 1)
frequencies = torch.exp(torch.arange(half, device=timesteps.device) * -scale)
angles = timesteps.float().unsqueeze(-1) * frequencies.unsqueeze(0)
return torch.cat((angles.sin(), angles.cos()), dim=-1)
```

</details>

More behaviour checks for this exercise: `check_sinusoidal_timestep_embedding` in [lesson12_checks.py](lesson12_checks.py).

````python
# Implement: Exercise 12.1
def sinusoidal_timestep_embedding(timesteps: torch.Tensor, dim: int) -> torch.Tensor:
    # (B,) -> (B, dim): [sin(t f_0) ... sin(t f_{h-1}), cos(t f_0) ... cos(t f_{h-1})], f_k = exp(-k log(10000)/(h-1)).
    if dim % 2 != 0:
        raise ValueError("dim is not even")

    n = dim // 2
    k = torch.arange(n, device=timesteps.device)
    f = torch.exp(-k * math.log(10_000) / (n - 1))[None, :]
    t = timesteps[:, None].float()

    e = torch.cat((torch.sin(t * f), torch.cos(t * f)), dim=-1)
    return e


# Check: Exercise 12.1
t = torch.tensor([0, 1, 7, 99])
for dim in (16, 32, 128):
    mine = sinusoidal_timestep_embedding(t, dim)
    assert mine.shape == (4, dim) and mine.dtype == torch.float32, f"Expected float32 (4, {dim})."
    close(
        mine,
        DiffusionSinusoidalPosEmb(dim)(t),
        f"dim {dim} differs from the pinned embedding",
        atol=1e-5,
    )
zero = sinusoidal_timestep_embedding(torch.tensor([0]), 8)
assert torch.equal(zero[0, :4], torch.zeros(4)) and torch.equal(zero[0, 4:], torch.ones(4)), (
    "Timestep 0 is all sines 0 then all cosines 1."
)
close(
    sinusoidal_timestep_embedding(torch.tensor([5]), 8)[0, 0],
    torch.tensor(math.sin(5.0)),
    "first frequency is 1",
    atol=1e-6,
)
close(
    sinusoidal_timestep_embedding(torch.tensor([5]), 8)[0, 3],
    torch.tensor(math.sin(5 * 1e-4)),
    "last frequency is 1/10000",
    atol=1e-6,
)
print("Exercise 12.1 passed")

# More behaviour checks: lesson12_checks.py, next to this notebook.
from lesson12_checks import check_sinusoidal_timestep_embedding

check_sinusoidal_timestep_embedding(sinusoidal_timestep_embedding)
````

The main branch applies the first convolution block, condition-derived scale and bias, then the second convolution block. Add the original input through an identity path, or a 1-by-1 convolution when channel widths differ. The full branching diagram is in the diffusion guide.


The main path learns a change to the input features. The shortcut carries the original features to the addition, which gives optimization a direct path through the block. When input and output channel counts match, the shortcut is the identity. When they differ, a learned 1×1 convolution changes the channel count. It preserves the time length. FiLM modulates the main path; it does not replace the shortcut.



## Exercise 12.2 — FiLM-conditioned residual block

**Purpose.** Write the block that is repeated through the whole U-Net: two convolutions, with the condition modulating the features between them.

**Contract (shapes, units, masks).**

- `block` has `conv1`, `conv2` (`Conv1dBlock`: Conv1d → GroupNorm → Mish, keeps the length), `cond_encoder` (`Mish → Linear(cond_dim, K)`), `residual_conv` (`Conv1d(in, out, 1)` or `Identity`), `out_channels` (`C`) and the bool `use_film_scale_modulation`.
- `x (B, C_in, T)` (channels first), `cond (B, cond_dim)`. Return `(B, C, T)`. Here $B$ is the batch size, $C_{in}$ and $C$ the input and output channel counts, $T$ the number of action-window slots (not the diffusion step count), and `cond_dim` the width of the condition vector.
- `out = conv1(x)`. `embed = cond_encoder(cond)` reshaped to `(B, K, 1)` so it broadcasts over `T`.
- **With scale modulation** (`K = 2C`): `scale = embed[:, :C]`, `bias = embed[:, C:]`, `out = scale * out + bias`. **Without** (`K = C`): `out = out + embed`.
- `out = conv2(out)`. Return `out + residual_conv(x)`.
- The pinned code uses `scale * out + bias`, not `(1 + scale) * out + bias`.

**Method and source.** FiLM, $\gamma(\text{cond})\odot F+\beta(\text{cond})$ from the background above ($\odot$ is element-wise multiplication) — [Perez et al. 2018](https://arxiv.org/abs/1709.07871). Diffusion Policy uses it in each 1D residual block ([paper site](https://diffusion-policy.cs.columbia.edu/)). Pinned class: `DiffusionConditionalResidualBlock1d`.

**Pointers.** Follow the order of the contract line by line. The only decision is how to cut `embed` into `scale` and `bias`. The residual connection adds the *input* (after `residual_conv`), not the output of `conv1`.

<details><summary>Detailed hint</summary>

```python
out = block.conv1(x)
embed = block.cond_encoder(cond).unsqueeze(-1)              # (B, K, 1)
if block.use_film_scale_modulation:
    scale, bias = embed[:, : block.out_channels], embed[:, block.out_channels :]
    out = scale * out + bias
else:
    out = out + embed
out = block.conv2(out)
return out + block.residual_conv(x)
```

</details>

More behaviour checks for this exercise: `check_film_residual_forward` in [lesson12_checks.py](lesson12_checks.py).

````python
# Implement: Exercise 12.2
def film_residual_forward(block: FilmParts, x: torch.Tensor, cond: torch.Tensor) -> torch.Tensor:
    # conv1 -> FiLM modulation from cond -> conv2 -> add residual_conv(x).
    skip = x

    x = block.conv1(x)
    embed = block.cond_encoder(cond).unsqueeze(-1)

    if block.use_film_scale_modulation:
        C = embed.shape[-2] // 2
        scale = embed[:, :C]
        bias = embed[:, C:]
        x = scale * x + bias
    else:
        x = x + embed

    x = block.conv2(x)
    x = x + block.residual_conv(skip)

    return x


# Check: Exercise 12.2
for scale in (True, False):
    torch.manual_seed(2)
    oracle = DiffusionConditionalResidualBlock1d(
        8, 16, 10, kernel_size=5, n_groups=8, use_film_scale_modulation=scale
    )
    mine = ConditionalResidualBlock1d(8, 16, 10, 5, 8, scale, forward_fn=film_residual_forward)
    mine.load_state_dict(oracle.state_dict(), strict=True)  # same layer names: identical weights
    x, cond = torch.randn(3, 8, 16, requires_grad=True), torch.randn(3, 10, requires_grad=True)
    out = mine(x, cond)
    assert out.shape == (3, 16, 16), "Expected (B, out_channels, T)."
    close(out, oracle(x, cond), f"output (scale modulation={scale}) differs from the pinned block")
    out.square().sum().backward()
    x2, cond2 = (t.detach().clone().requires_grad_(True) for t in (x, cond))
    oracle(x2, cond2).square().sum().backward()
    close(x.grad, x2.grad, f"input gradient (scale={scale})", atol=1e-5)
    close(cond.grad, cond2.grad, f"condition gradient (scale={scale})", atol=1e-5)
    close(
        mine.cond_encoder[1].weight.grad,
        oracle.cond_encoder[1].weight.grad,
        "FiLM layer gradient",
        atol=1e-5,
    )
    assert not torch.allclose(mine(x, cond + 1.0), out.detach(), atol=1e-4), (
        "The condition must change the output."
    )
same = ConditionalResidualBlock1d(16, 16, 10, 5, 8, True, forward_fn=film_residual_forward)
assert isinstance(same.residual_conv, nn.Identity) and same(
    torch.randn(2, 16, 16), torch.randn(2, 10)
).shape == (2, 16, 16), "Also check equal channels."
print("Exercise 12.2 passed")

# More behaviour checks: lesson12_checks.py, next to this notebook.
from lesson12_checks import check_film_residual_forward

check_film_residual_forward(film_residual_forward)
````
